```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('housing.csv')
```

```python
df.head()
```

```python
#Q1
plt.figure(figsize=(15,8))
sns.scatterplot(data=df, x='Landsize', y='Price', hue='Type')
```

```python
average_price=df.groupby('YearBuilt')['Price'].mean()
```

```python
#Q3
sns.lineplot(x=average_price.index, y=average_price)
```

```python
region_average= df.groupby('Regionname')['Price'].mean()
```

```python
#Q5
plt.figure(figsize=(20,5))
sns.barplot( x=region_average.index, y=region_average)
```

```python
#Q7
sns.histplot(data=df, x='Price')
plt.ylabel('Properties')
```

```python
#Q9
sns.boxplot(data=df, x='YearBuilt', y='Regionname')
plt.ylabel('Region')
plt.xlabel('Year Built')
```
